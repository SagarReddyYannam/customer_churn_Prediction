```python
X_train=pd.get_dummies(X_train,drop_first=True)
X_test=pd.get_dummies(X_test,drop_first=True) # converts text categories into numerical columns
#get_dummies() performs one hot encoding,it converts categories into 0 and 1
#drop_first=True----this removes one category from each categorical column
```

```python
X_test=X_test.fillna(X_train.median(numeric_only=True)) #fills the missing value in X_test using the median from X_train
#numeric_only=True-- means only numeric columns are considered
#fillna() means fill the missing values NaN and replaces that missing value with the appropriate median
```

Feature Scaling

```python
from sklearn.preprocessing import StandardScaler  ##import standardscaler for scaling numerical features
##StandardScaler --it changes numerical values to a common scale
```

```python
scaler=StandardScaler()# creates a StandardScaler object that we will use for scale for features
X_train_scaled=scaler.fit_transform(X_train) #learns the scaling parameters from X_train and then scales X_train
#fit() learns from the training data
#transform() uses those values to scale the data
```

```python
X_test_scaled=scaler.transform(X_test) #scales X_test using the same scaling learned fron X_train-
#X_train--learn the scaling + scale it
#X_test--only apply the learned scaling
```

```python
X_train_scaled.shape #shows the number of rows and columns in the scaled training data
```

```python
X_test_scaled.shape #shows the number of rows and columns in the scaled test data
```

Train our first model

```python
from sklearn.linear_model import LogisticRegression #import LogisticRegression for predicting customer churn
```

```python
model=LogisticRegression(max_iter=1000) #creates a LogisticRegression model
```

```python
model.fit(X_train_scaled,y_train) #trains the model using training data
```

Make Predictions

```python
y_pred=model.predict(X_test_scaled) #predicts churn for the test customers
```

Check how good the predictions are

```python
from sklearn.metrics import accuracy_score #imports the accuracy calculations
accuracy=accuracy_score(y_test,y_pred) #compares actual values with predicted values
```

```python
accuracy #dispalys the accuracy
```

Create a final prediction function

```python
#instead of manually writing all 33 values every time ,we will create a function where we can give customer details and get the churn prediction

def predict_churn(customer_data):

    customer_scaled = scaler.transform(customer_data)

    prediction = model.predict(customer_scaled)

    probability = model.predict_proba(customer_scaled)[0][1]

    result = prediction[0]

    print("Churn Prediction:", result)
    print("Churn Probability:", probability)

    return result, probability
```
